## Infrared Arrays: 

- first IR array use 1984/1985; relatively high read out noise 
    - IR developments driven mostly from military needs
    - FIR/radio from astronomers


- Can dope Si so it is sensitive in visible and still act as CCD
- Need other materials to be sensitive at long wavelengths
  - InSb & Hg$_{1-x}$Cd$_{x}$Te   $\rightarrow \lambda < 5 \mu$m
  - Hg$_{0.55}$Cd$_{0.45}$Te $\rightarrow \lambda <  2.4 \mu$m
  - HgCdTe $\rightarrow \lambda < 10 \mu$m (used for JWST NIRCam (https://jwst-docs.stsci.edu/jwst-near-infrared-camera#gsc.tab=0)
  - Si:As  $\rightarrow \lambda < 28 \mu$m (used in Spitzer and JWST MIRI)
  - Si:Sb  $\rightarrow \lambda < 40 \mu$m (used in Spitzer spectrograph)
  - HgCdTe QE curves reach >80% in the IR (https://web.archive.org/web/20180725212841id_/http://iopscience.iop.org/article/10.1086/590161/pdf; https://www.spiedigitallibrary.org/conference-proceedings-of-spie/11180/111803D/Teledynes-high-performance-infrared-detectors-for-Space-missions/10.1117/12.2536040.full?SSO=1); 

- Hybrid construction:
    - "Background limited" sources fill wells fast, so need to read out up to 1 kHz
    - CCD takes several seconds to read
    - Solution: IR absorbing layer connected to Si multiplexer
    
    - amplifiers are grown on a thin wafer of silicon with a grid of input contacts, one for each amplifier 
    - detectors are grown on a wafer of suitable semiconductor, with a matching grid of output contacts
    - Indium bumps are deposited on each grid of contacts and connect the detector and amplifier grids 
    - Major problems: thermal contraction mismatch, can crack

 <img src="indium_arrays.png" width=400 height=400 /> (Rieke, 2007)


- Each unit cell reads out directly in turn to an amplifier: no charge motion across chip

- Often have many amplifiers (~32), each reading quadrants or every 4th row, etc.
    - trading readout speed against power dissipation as # increases
- Huge read noise, up to 2000 $e^{-}$
    - if *extremely* controlled env. can be as low as 4, but not-feasible IRL$^*$
- Can measure voltage on each unit cell diode more than once: non-destructive reads (NDR)
- Possible to select subarray and read it out independently $\rightarrow$ signal sent to independent amplifier 
- Correlated multiple sampling reduces read noise by $\sqrt{\mbox{NDR}}$


- Trade-off between read speed and noise: CCD slow and accurate, multiplexer fast and noisy



##  Infrared Arrays vs CCDs: 
- More susceptible to dark current (but give you access to the long $\lambda$s)
- Also, camera has actual thermal emission
    - cool to 4 K with liquid He
    - QE ~ same as CCDs
        - but, QE varies more from pixel to pixel 
    - More bad pixels/regions than CCDs (bumps separate, cracks, non-Si technology lags)
    - Same additive and multiplicative effects in multiplexer as with CCD, but worse
    - Systematics have much more drift during night
    - Also need to correct background.  Sky+telescope+camera are very bright, collectively called ``sky''.
- IR array correction formula:

<img src="irarray_correction.png" width=400 height=400 /> 


- $S$ is sky image taken close to exposure, includes bias, dark current, telescope, camera, etc.
- Different techniques for getting flats and sky, depending on $\lambda$



### Summary of array effects:
  - Sky background (+, smooth, time-variable)
  - Dark current (+, smooth and not, mostly steady)
  - Biases from multiple amplifiers (+, not smooth, steady)
  - QE ($\times$, not smooth, steady)
  - Cosmic rays (X, not smooth, time-variable)
  - Bad pixels (X, not smooth, steady)
$$ADU = g_{2} ( g_{1} ( E_{\rm Q} ( F_{*} + F_{\rm sky} ) + I_{\rm d} )+ V_{\rm off} )$$


$$ ADU =  g_{2} g_{1} E_{\rm Q} F_{*} 
     + g_{2} g_{1} E_{\rm Q} F_{\rm sky}
     + g_{2} g_{1} I_{\rm d}
     + g_{2} V_{\rm off}$$

#### where:
   - $g_{1}$ Volt/electron from unit cell
   - $g_{2}$ ADU/Volt from array amp + ADC
   - $E_{\rm Q}$ quantum efficiency
   - $F$ light fluxes
   - $I_{\rm d}$ dark current
   - $V_{\rm off}$ bias voltage on unit cell gate
 
####  
- Single image: all array effects present, many vary spatially, often
     w/same appearance as real objects
- Stars, galaxies, planets also vary spatially
- Must move completely off objects for sky frame to disentangle signal from noise
- If sources take much of frame, nod off frame
- If sources small, nod can be small and objects can still be on frame: dithering (jittering in spectroscopy)


- No time "wasted" getting sky since all frames have objects 
- Stars/galaxies/etc. shift to different pixels, steady
- Noise/systematics stay in place, may average out in time
