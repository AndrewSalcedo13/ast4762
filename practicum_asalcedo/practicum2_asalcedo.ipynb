```python
#Problem 1: Write a code that creates a subsample containing N random draws
#from a Gaussian distribution with a certain width and mean
def random(mu, sigma, N):
    samples = np.random.normal(mu, sigma, N)
    return samples
samples_i = random(55, 13, 10)
print(samples_i)
```

```python
#Problem 2: create a sample with N=10 random draws
outputs = []

for i in range(10):
    samples_i = random(55, 13, 10)
    xbar = np.mean(samples_i)
    sigma = np.std(samples_i)
    outputs_i = [i, xbar, sigma]
    outputs.append(outputs_i)
outputs
```

```python
import numpy as np

def make_sample(N, mu=55, sigma=13, seed=None):
    """Draw N random samples from a Gaussian(mu, sigma) distribution."""
    rng = np.random.default_rng(seed)
    return rng.normal(loc=mu, scale=sigma, size=N)

N = 10
n_trials = 10
results = np.zeros((n_trials, 3))

for i in range(n_trials):
    sample = make_sample(N)                  # fresh random draw each iteration
    results[i, 0] = i
    results[i, 1] = np.mean(sample)
    results[i, 2] = np.std(sample, ddof=1)    # ddof=1 -> sample std dev (N-1 denominator)
```
